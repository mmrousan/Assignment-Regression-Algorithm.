```python
import pandas as pd
```

```python
dataset=pd.read_csv("insurance_pre.csv")
```

```python
#Since there are categorical data, we have convert them into numerical data
```

```python
dataset=pd.get_dummies(dataset,drop_first=True)
```

```python
#Now we are going to segragate Input and Output Variables
```

```python
dataset.columns
```

```python
independent=dataset[['age', 'bmi', 'children', 'sex_male', 'smoker_yes']]
```

```python
dependent=dataset[["charges"]]
```

```python
#Splitting the training set and test set
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(independent,dependent,test_size=0.30,random_state=0)
```

```python
#Without standardization- we got the result r2 score as =0.8950779234313312, so we came back at this point
#Trying to standardize whether we can improve our result
```

```python
#See the value of X_train
```

```python
#Once we completed splitting the training and test set, next is model creation
```

```python
from sklearn.ensemble import RandomForestRegressor
regressor=RandomForestRegressor(n_estimators=100, criterion="absolute_error", max_depth=None, min_samples_split=2, min_samples_leaf=1, max_features=0.7, bootstrap=True, random_state=0)

#Both lines of codes are just a procedure
#In kernal- what we are trying to define is whether using linear or non-linear

regressor=regressor.fit(X_train,y_train)
#model going to create
```

```python
y_pred=regressor.predict(X_test)
```

```python
from sklearn.metrics import r2_score
r_score=r2_score(y_test,y_pred)
```
